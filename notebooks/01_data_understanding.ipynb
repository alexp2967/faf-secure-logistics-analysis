```python
from pathlib import Path
import pandas as pd

csv_path = Path("/Users/sparrkie/faf-secure-logistics-analysis/data/raw/extracted/FAF5.7.1.csv")
df = pd.read_csv(csv_path)

df.head()
```

```python
# Create smaller working dataframe

id_cols = ["dms_orig", "dms_dest", "dms_mode", "sctg2", "trade_type", "dist_band"]

tons_cols = [f"tons_{year}" for year in range(2018, 2025)]
value_cols = [f"value_{year}" for year in range(2018, 2025)]
tmiles_cols = [f"tmiles_{year}" for year in range(2018, 2025)]

keep_cols = id_cols + tons_cols + value_cols + tmiles_cols

faf = df[keep_cols].copy()

faf.head()
```

```python
tons_col = [col for col in df.columns if "tons" in col]
val_col = [col for col in df.columns if "value" in col]
print(tons_col)
print(val_col)
```

```python
id_cols = ["dms_orig", "dms_dest", "dms_mode", "sctg2", "trade_type", "dist_band"]

records = []

for year in range(2018, 2025):
    temp = faf[id_cols].copy()
    temp["year"] = year
    temp["tons"] = faf[f"tons_{year}"]
    temp["value"] = faf[f"value_{year}"]
    temp["tmiles"] = faf[f"tmiles_{year}"]
    records.append(temp)

faf_long = pd.concat(records, ignore_index=True)
```

```python
faf_long.head()
faf_long.info()
```

```python
faf_long.to_csv("../data/processed/faf_long_2018_2024.csv", index=False)
```

```python
faf_long["dms_orig"] = faf_long["dms_orig"].astype("int32")
faf_long["dms_dest"] = faf_long["dms_dest"].astype("int32")
faf_long["dms_mode"] = faf_long["dms_mode"].astype("int8")
faf_long["sctg2"] = faf_long["sctg2"].astype("int8")
faf_long["trade_type"] = faf_long["trade_type"].astype("int8")
faf_long["dist_band"] = faf_long["dist_band"].astype("int8")
faf_long["year"] = faf_long["year"].astype("int16")

faf_long["tons"] = faf_long["tons"].astype("float32")
faf_long["value"] = faf_long["value"].astype("float32")
faf_long["tmiles"] = faf_long["tmiles"].astype("float32")
```
